# BoW Problemi

- "python" 10 kez "ve" 100 kez geçiyor. BoW'da "ve" daha önemli görünür

- Aslında "ve" anlamsız -- her yerde geçer.

- Çözüm: Yaygınlığı cezalandır, nadir ama önemli olanı öne çıkar.

# TF - Term Frequency

- Bu belgede kelimenin sıklığı

- TF = (kelime sayısı) / (belgedeki toplam kelime)

- Örnek "python" 3 kez, toplam 50 kelime -> TF = 0.06

# IDF - Inverse Document Frequency

- Kelimenin TÜM belgelerdeki nadirliği

- IDF = log(N / df)

- N = toplam belge, df = kelimenin geçtiği belge sayısı

- "ve" her belgede -> IDF düşük (log(N/N) = 0)

- "quantum" 5 belgede (N = 1000) -> IDF = log(1000%5) = 5.3

# TF X IDF -- Birleştir

- Skor = TF x IDF

# Özet

- TF = belgedeki sıklık

- IDF = nadir olma olasılığı

- TF x IDF = yaygın ama nadir olan ağırlıklı

- Arama motoru klasiği, RAG'da hala kullanılır

```python
# MANUEL TF-IDF SEZGI
docs = [
    "python makina öğrenmesi ve nlp",
    "python veri bilimi analiz",
    "veri bilimi istatistik"
]
```

```python
# SKLEARN TfidfVectorizer
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(docs)

df_tfidf = pd.DataFrame(
    X_tfidf.toarray(),
    columns= tfidf.get_feature_names_out(),
    index = [f"Belge {i + 1}" for i in range(len(docs))]
).round(3)
print("TF-IDF Matrix:")
print(df_tfidf)
```

```python
# BoW vs TF-IDF karşılaştırma
from sklearn.feature_extraction.text import CountVectorizer

bow = CountVectorizer()
X_bow = bow.fit_transform(docs)

df_bow = pd.DataFrame(
    X_bow.toarray(),
    columns= bow.get_feature_names_out(),
    index = [f"Belge {i + 1}" for i in range(len(docs))]
)
print("\nBoW (sayım)")
print(df_bow)
print("\nTF-IDF (ağırlıklı):")
print(df_tfidf)

# BoW'da 1, 1 olan degerler TF-IDF'de farklı - nadir olan daha yüksek
```

# CountVectorizer -> TfidfVectorizer

Parametreler --> Aynı Mı?

- max_features, min_df, max_df  --> Aynı
- ngram_range, stop_words --> Aynı
- lowercase, token_pattern --> Aynı

- norm --> "l2" (default, cosine sim için ideal)
- sublinear_tf --> Tarue --> log(1+TF) -- büyük belgelerde iyi
- smooth_idf --> True --> sıfır bölme önle

```python
# Her Belge için en yüksek TF-IDF skorlu Kelimeler
feature_names = tfidf.get_feature_names_out()

# Belge 1 
scores = X_tfidf[0].toarray().flatten()
top_idx = scores.argsort()[::-1][:3]
print(f"\nBelge 1: {docs[0]}")
print(f"    {feature_names[top_idx[0]]}: {scores[top_idx[0]]:.3f}")
print(f"    {feature_names[top_idx[1]]}: {scores[top_idx[1]]:.3f}")
print(f"    {feature_names[top_idx[2]]}: {scores[top_idx[2]]:.3f}")

# Belge 2
scores = X_tfidf[1].toarray().flatten()
top_idx = scores.argsort()[::-1][:3]
print(f"\nBelge 1: {docs[1]}")
print(f"    {feature_names[top_idx[0]]}: {scores[top_idx[0]]:.3f}")
print(f"    {feature_names[top_idx[1]]}: {scores[top_idx[1]]:.3f}")
print(f"    {feature_names[top_idx[2]]}: {scores[top_idx[2]]:.3f}")

# Belge 3
scores = X_tfidf[2].toarray().flatten()
top_idx = scores.argsort()[::-1][:3]
print(f"\nBelge 1: {docs[2]}")
print(f"    {feature_names[top_idx[0]]}: {scores[top_idx[0]]:.3f}")
print(f"    {feature_names[top_idx[1]]}: {scores[top_idx[1]]:.3f}")
print(f"    {feature_names[top_idx[2]]}: {scores[top_idx[2]]:.3f}")
```

# Cosine Similarity

- TF-IDF vektörleri arasında benzerlik

from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(X[0], X[1])

- [[0.0246]] -- İki belge ne kadar benzer (0-1)

- Arama motorunun özü: sorgu --> TF-IDF --> en yakın belgeleri bul

- RAG'da vektör DB kullanılmadan önce basit retrieval için hala geçerli 

```python
# Belge BENZERLİĞİ -- Cosine Similarity
# TfidfVectorizer çıktısı L2-normalize -> cosine similarity = vektörlerin nokta çarpması.
# "Hangi iki belge birbirine en çok benziyor?" sorusu: arama motoru & RAG'ın temeli

import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

sim = cosine_similarity(X_tfidf)
sim_df = pd.DataFrame(
    sim.round(3),
    columns= [f"Belge {i+1}" for i in range(len(docs))],
    index= [f"Belge {i+1}" for i in range(len(docs))]
)

print("Belgeler")
print(f"    Belge 1: {docs[0]}")
print(f"    Belge 1: {docs[1]}")
print(f"    Belge 1: {docs[2]}")
print("\nCosine similarity matrisi")
print(sim_df)

# En benzer çifti bul (köşegen = bölgenin kendisi, yok say)
sim_no_diag = sim.copy()
np.fill_diagonal(sim_no_diag,0)
i,j = np.unravel_index(sim_no_diag.argmax(), sim_no_diag.shape)
print(f"\nEn benzer çift: Belge {i+1} <--> Belge {j + 1} (similarity = {sim_no_diag[i,j]:.3f})")
print("-> Ortak kelimeleri paylaştıklaro için yakınlar.")
print("\n Arama motorları & RAG (ML-08) tam bunu yapar:")
print("     sorgu da TF-IDF vektörü olur -> en yüksek cosine'li belgeleri getir.")
```

# Yan Yana Karşılaştırma

- BoW (CountVectorizer)  -----------    TF-IDF

- Değer - Sayım(tam sayı)               Ağırlık (ondalık)

- Yaygın kelime      Baskın              Zayıflatılıe

- Nadir Kelime      Uyumlu (sayım ister)    Olur ama BoW daha iyi

- LogReg/SVM        Kabul edilebilir        Çok iyi

- Hız           Çok hızlı           Hızlı

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

categories = ["rec.autos","sci.med","rec.sport.hockey"]
new_train = fetch_20newsgroups(subset= 'train', categories= categories)
new_test = fetch_20newsgroups(subset= 'test', categories= categories)

pipelines = {
    'BoW + NB': Pipeline([('v', CountVectorizer(max_features = 5000)), ('m', MultinomialNB())]),
    'TF-IDF + NB': Pipeline([('v', TfidfVectorizer(max_features=5000)), ("m", MultinomialNB())]),
    'TF-IDF + LogReg': Pipeline([('v', TfidfVectorizer(max_features= 5000)), ('m',LogisticRegression(max_iter=100))]) 
}

print(f" {'Pİpeline':<20} {'Accuracy':>10}")
print("-" * 35)
for name, pipe in pipelines.items():
    pipe.fit(new_train.data, new_train.target)
    acc = pipe.score(new_test.data, new_test.target)
    print(f"{name:<20} {acc:>10.2%}")
```

```python
# TF-IDF için TİPİK PRODUCTİIN PIPELINE
# Vektörleştirme + model TEK objede. Üretimde tam oalrak böyle kullanılır.
from sklearn.model_selection import cross_val_score

optimal_pipeline = Pipeline([
    ("vec", TfidfVectorizer(
        max_features = 10_000,
        min_df = 3,
        max_df = 0.8,
        ngram_range= (1,2),
        sublinear_tf= True,
        stop_words = 'english'
    )),
    ("clf", LogisticRegression(max_iter= 1000, C = 1))
])

# 1) Croos-Validation ile güvenirliği ölç
scores = cross_val_score(optimal_pipeline, new_train.data, new_train.target, cv = 3)
print(f"Pipeline CV doğruluğu: {scores.mean():.3f} + {scores.std():.3f}")

# 2) Tüm eğitim verisiyele fit et -> artık obje, production'a hazır
optimal_pipeline.fit(new_train.data, new_train.target)

# 3) Ham bir metni doğrudan tahmin et (vektörleştirme pipeline içinde otomatik yapılır.)
ornek = "The new turbo engine boosts horsepower and torque"
tahmin = optimal_pipeline.predict([ornek])[0]
print(f"\nGirdi : {ornek}")
print(f"Tahmin: {new_train.target_names[tahmin]}")

# (Deploy: pickle.dump(optimal_pipeline,...) -> tek dosya, predict ile kullan)
```
