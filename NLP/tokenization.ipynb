# Token Nedir

En Küçük anlamlı birim - Kelime, ek, hatta karakter

- Kelime: ["Bugün", "NLP", "öğreniyorum"]
- Subword: ["Bugün", "NLP", "öğren",  "##iyor", "##um"]
- Karakter: ["B","u","g","u","n", ...]

## Tokenization Seviyeleri

- Kelime ---> BoW, TF-IDF, klasik NLP

- Cumle ---> Özetleme, NEW, QA sistemler

- Subword (BPE/WordPiece) ---> Modern LLM (GPT.BERT)

- Karakter ---> Yazım düzeltme, düşük kaynak diller

## BASİT TOKENİZATİON YETMEZ

1. text = "Dr. Smith'in PhD tezi 2024'te yayınlandı. Harika!"

2. text.split()

3. ['Dr.','Smith\in','PhD', 'tezi', '2024\te','yayınlandı.','Harika']

- "Dr." sonunda cümle bitişi sayılır

- "Smith'in" apostrof tek token mı?

- "Harika" noktalama bitişik

```python
import nltk
nltk.download("punkt", quiet= True) # cümle/kelime tokonizer verisi (word_tokenize için)
nltk.download("stopwords", quiet= True) # durak kelime listeleri (the, and , is, ...) - filtrelemek için
nltk.download("wordnet", quiet= True) # lemmatization sözlüğü (WordNetLemmatizer için kök bilgisi)
nltk.download("averaged_perceptron_tagger", quiet= True) # POS tagger verisi (kelime türü: isim/fiil...)
```

```python
# TOKENIZATION 
from nltk.tokenize import word_tokenize, sent_tokenize

text = "Dr. Smith'in PhD tezi NLP üzerineydi. Harika bir çalışmaydı."

# Cümle bazlı
sentences = sent_tokenize(text,language= "english")
print(f"  -  {sentences[0]} ")
print(f"  -  {sentences[1]} ")

# Kelime Bazlı
words = word_tokenize(text, language= "english")
print(f"\nKelimeler: {words}" )
```

```python
# BASİT .split() neden yetmez?
simple = text.split()
print("\nSplit:", simple)
```
