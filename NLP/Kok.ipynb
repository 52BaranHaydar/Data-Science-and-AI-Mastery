# Problem: Kelime Çeşitliliği

- run, runs, running, ran -> aynı "koşmak"

- koşuyordum, koştum, koşacak -> aynı "koş"

Çözüm: hepsini tek köke indir -> vocabulary küçülür, model güçlenir

# Stemming - Kaba Kök

 - Kural tabanlı son ek kesme 
 - Hızlı ama bazen dilbilim olarak yanlış

```python
import nltk
nltk.download("punkt", quiet= True) # cümle/kelime tokonizer verisi (word_tokenize için)
nltk.download("stopwords", quiet= True) # durak kelime listeleri (the, and , is, ...) - filtrelemek için
nltk.download("wordnet", quiet= True) # lemmatization sözlüğü (WordNetLemmatizer için kök bilgisi)
nltk.download("averaged_perceptron_tagger", quiet= True) # POS tagger verisi (kelime türü: isim/fiil...)
```

```python
# STEMMING - kaba kök
from nltk.stem import PorterStemmer, SnowballStemmer

porter = PorterStemmer()
snowball = SnowballStemmer("english")

words = ["running", "runs","ran","better", "studies", "studying", "played"]
print(f"{'Kelime':<12} {'Porter':<10} {'Snowball':<10}")
print("-" * 35)
print(f"{'running':12} {porter.stem('running'):<10} {snowball.stem('running'):<10}")
print(f"{'runs':12} {porter.stem('runs'):<10} {snowball.stem('runs'):<10}")
print(f"{'ran':12} {porter.stem('ran'):<10} {snowball.stem('ran'):<10}")
print(f"{'better':12} {porter.stem('better'):<10} {snowball.stem('better'):<10}")
print(f"{'studies':12} {porter.stem('studies'):<10} {snowball.stem('studies'):<10}")
print(f"{'studting':12} {porter.stem('studying'):<10} {snowball.stem('studying'):<10}")
print(f"{'played':12} {porter.stem('played'):<10} {snowball.stem('played'):<10}")
```

# LEMMATIZATION - DOGRU Kök
- Sözlük tabanlı dilbilim analizi

- Yavaş ama doğru. POS tag gerekir (verb, noun, adjective)

```python
# LEMMATIZATION - dilbilimsel doğru kök
from nltk.stem import WordNetLemmatizer

lemma = WordNetLemmatizer()

# Varsayılan: noun
print(f"\nLemmatization (POS bazlı)")
print(f" running (verb): {lemma.lemmatize('running', pos = "v")} ") # run
print(f" running (noun): {lemma.lemmatize('running', pos = "n")} ") # running
print(f" better (verb): {lemma.lemmatize('better', pos = "a")} ") # good
print(f" studies (verb): {lemma.lemmatize('studies', pos = "v")} ") # study

# Stemming vs Lemmatiation karşılaştır
print(f"\n{'Word':<12} {'Stemmer':<10} {'Lemmatizer':<12}")
print("-" * 37)
print(f"{'running':<12} {porter.stem('running'):<10} {lemma.lemmatize('running'):<10}")
print(f"{'runs':<12} {porter.stem('runs'):<10} {lemma.lemmatize('runs'):<10}")
print(f"{'ran':<12} {porter.stem('ran'):<10} {lemma.lemmatize('ran'):<10}")
print(f"{'better':<12} {porter.stem('better'):<10} {lemma.lemmatize('better'):<10}")
print(f"{'studies':<12} {porter.stem('studies'):<10} {lemma.lemmatize('studies'):<10}")
print(f"{'studting':<12} {porter.stem('studying'):<10} {lemma.lemmatize('studying'):<10}")
print(f"{'played':<12} {porter.stem('played'):<10} {lemma.lemmatize('played'):<10}")

```
