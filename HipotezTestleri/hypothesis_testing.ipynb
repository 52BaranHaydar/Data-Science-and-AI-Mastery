# 🧪 Hipotez Testleri ve İstatistiksel Karar Alma (Hypothesis Testing)

Bu çalışmada, veri biliminde ve A/B testlerinde karar alma süreçlerinin kalbini oluşturan **Hipotez Testleri** ele alınmaktadır.

---

## 📌 1. Temel Kavramlar & Karar Mekanizması

### 🎯 Hipotez Nedir?
- **Sıfır Hipotezi ($H_0$ - Null Hypothesis):** Değişimin, etkinin veya farkın olmadığını, durumun tesadüfi olduğunu savunan varsayımdır (Örn: *"Yeni tasarım satışları etkilemedi, $ \mu = \mu_0 $"*).
- **Alternatif Hipotez ($H_1$ veya $H_a$ - Alternative Hypothesis):** Araştırmacının kanıtlamaya çalıştığı, istatistiksel bir farkın veya etkinin var olduğunu savunan varsayımdır (Örn: *"Yeni tasarım satışları artırdı, $ \mu > \mu_0 $"*).

### ⚖️ Karar Kriterleri:
- **Anlamlılık Düzeyi (Significance Level - $\alpha$):** Genellikle $\alpha = 0.05$ (%5 risk) kabul edilir. $H_0$ doğruyken onu yanlışlıkla reddetme (Tip 1 Hata) olasılığıdır.
- **p-değeri (p-value):** $H_0$ doğru kabul edildiğinde, elimizdeki örneklem kadar veya daha aşırı bir sonucun şans eseri elde edilme olasılığıdır.
  - 🟢 **$p < \alpha$:** $H_0$ **reddedilir**. Fark istatistiksel olarak **anlamlıdır** (şans eseri değildir).
  - 🔴 **$p \ge \alpha$:** $H_0$ **reddedilemez**. Fark istatistiksel olarak **anlamlı değildir** (tesadüfi olabilir).

---

## 🧭 Z-Testi vs T-Testi Karar Tablosu

| Kriter | Z-Testi | T-Testi |
| :--- | :--- | :--- |
| **Popülasyon Standart Sapması ($\sigma$)** | Biliniyor | **Bilinmiyor** (Örneklem $s$ kullanılır) |
| **Örneklem Büyüklüğü ($n$)** | Genellikle $n \ge 30$ | Hem $n < 30$ hem de büyük örneklemler |
| **Test İstatistiği** | $Z = rac{ar{X} - \mu_0}{\sigma / \sqrt{n}}$ | $t = rac{ar{X} - \mu_0}{s / \sqrt{n}}$ |
| **Dağılım** | Standart Normal Dağılım $N(0, 1)$ | Student-t Dağılımı ($df = n - 1$) |


---
## 🎯 2. Tek Örneklem Z-Testi (One-Sample Z-Test)

### 📐 Formül:
$$Z = \frac{\bar{X} - \mu_0}{\frac{\sigma}{\sqrt{n}}}$$

- $\bar{X}$: Örneklem ortalaması
- $\mu_0$: Popülasyon ortalaması (iddia edilen değer)
- $\sigma$: Popülasyon standart sapması (biliniyor olmalı)
- $n$: Örneklem sayısı ($n \ge 30$)

#### 💡 Örnek Senaryo 1: Şişe Su Fabrikası Dolum Kontrolü
Fabrikada ortalama $500\text{ ml}$ dolum yapılmaktadır. Popülasyon $\sigma = 10\text{ ml}$. Alınan $36$ şişelik örneklemin ortalaması $495\text{ ml}$ çıkmıştır. Fark anlamlı mıdır?


```python
from scipy.stats import norm
import math

# Popülasyon bilgileri
pop_mean = 500
pop_std = 10

# Örneklem bilgileri
n = 36
sample_mean = 495

# Z istatistiği hesabı
z = (sample_mean - pop_mean) / (pop_std / math.sqrt(n))

# İki yönlü p-değeri
p_value = 2 * (1 - norm.cdf(abs(z)))

alpha = 0.05
print(f"Z İstatistiği: {z:.4f}")
print(f"p-değeri     : {p_value:.6f}")

if p_value < alpha:
    print(" Sonuç: H0 reddedilir! Ortalama 500 ml den istatistiksel olarak anlamlı şekilde farklıdır.")
else:
    print(" Sonuç: H0 reddedilemez. Fark şans eseri oluşmuş olabilir.")
```

#### 💡 Örnek Senaryo 2: IT Şirketinde Çevik (Agile) Dönüşüm
Bir yazılım firmasında geliştiricilerin ortalama kod teslim süresi $\mu = 10$ gündür ($\sigma = 2.5$). 
Yeni uygulanan Agile yöntem sonrası $40$ kişilik ekibin teslim süresi $ar{X} = 9.4$ gün olarak ölçülmüştür.
Yeni yöntem teslim süresini anlamlı şekilde düşürdü mü? (%95 güven düzeyi, $\alpha = 0.05$)


```python
from scipy.stats import norm
import math

pop_mean = 10
pop_std = 2.5
n = 40
sample_mean = 9.4

z = (sample_mean - pop_mean) / (pop_std / math.sqrt(n))
p_value = 2 * (1 - norm.cdf(abs(z)))

alpha = 0.05
print(f"Z İstatistiği: {z:.4f}")
print(f"p-değeri     : {p_value:.4f}")

if p_value < alpha:
    print(" Sonuç: H0 reddedilir. Agile yöntem teslim süresini istatistiksel olarak anlamlı şekilde değiştirmiştir.")
else:
    print(" Sonuç: H0 reddedilemez. Değişim tesadüfi olabilir.")
```

## 📊 3. Tek Örneklem T-Testi (One-Sample T-Test)

Gerçek hayat problemlerinde popülasyon standart sapması $\sigma$ genellikle **bilinmez**. 
Bunun yerine örneklemin kendi standart sapması $s$ hesaplanır ve **Student-t dağılımı** kullanılır.

### 📐 Formül:
$$t = \frac{\bar{X} - \mu_0}{\frac{s}{\sqrt{n}}}, \quad df = n - 1$$

#### 💡 Örnek Senaryo 3: Ürün Satış Ortalaması Testi
Geçmiş satış ortalaması $\mu = 50$ adet olan bir ürünün yeni dönemdeki 10 günlük satışları: `[52, 49, 51, 50, 53, 54, 48, 49, 50, 52]`.
Popülasyon varyansı bilinmediği için T-testi uygulanır.


```python
import numpy as np
from scipy.stats import ttest_1samp

# Veriler
mu = 50
data = np.array([52, 49, 51, 50, 53, 54, 48, 49, 50, 52])

t_stat, p_value = ttest_1samp(data, popmean=mu)

print(f"Örneklem Ortalaması: {data.mean():.2f}, Standart Sapması (s): {data.std(ddof=1):.2f}")
print(f"T-İstatistiği: {t_stat:.4f}")
print(f"p-değeri     : {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print(" Sonuç: H0 reddedilir. Satış ortalamasında anlamlı bir değişim var.")
else:
    print(" Sonuç: H0 reddedilemez. Satış ortalamasındaki fark şans eseridir.")
```

#### 💡 Örnek Senaryo 4: Kahve Zinciri Porsiyon Kontrolü
Bir kahve zinciri, bardak kahvelerin ortalama $250\text{ ml}$ olduğunu iddia etmektedir. 
Müşteri şikayetleri üzerine 15 bardak kahve ölçülmüştür: `[248, 252, 251, 249, 247, 253, 250, 248, 251, 249, 250, 252, 248, 247, 251]`
Ortalama iddia edildiği gibi $250\text{ ml}$ midir?


```python
import numpy as np
from scipy.stats import ttest_1samp

data = np.array([248, 252, 251, 249, 247, 253, 250, 248, 251, 249, 250, 252, 248, 247, 251])
pop_mean = 250

t_stat, p_value = ttest_1samp(data, popmean=pop_mean)

print(f"Örneklem Ortalaması: {data.mean():.2f} ml (n={len(data)})")
print(f"T İstatistiği: {t_stat:.4f}")
print(f"p-değeri     : {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print(" Sonuç: H0 reddedilir. Ortalama 250 ml den anlamlı derecede farklıdır.")
else:
    print(" Sonuç: H0 reddedilemez. Kahve miktarı 250 ml standardına uygundur.")
```

#### 💡 Örnek Senaryo 5: İlaç Etkinliği Simülasyonu
Bir ilaç firması yeni geliştirdiği formülün uyku süresini artırdığını iddia etmektedir.
- Eski ortalama: $6.0$ saat
- $25$ kişilik denek grubunun simüle edilen uyku süreleri ($ar{X} pprox 6.4$, $s pprox 0.6$).


```python
import numpy as np
from scipy.stats import ttest_1samp

np.random.seed(42) # Tekrarlanabilirlik için
pop_mean = 6.0
data = np.random.normal(loc=6.4, scale=0.6, size=25)

t_stat, p_value = ttest_1samp(data, popmean=pop_mean)

print(f"Gözlenen Ortalama Uyku Süresi: {data.mean():.2f} saat")
print(f"T-istatistiği: {t_stat:.4f}")
print(f"p-değeri     : {p_value:.6f}")

alpha = 0.05
if p_value < alpha:
    print(" Sonuç: H0 reddedilir! Yeni ilaç uyku süresini istatistiksel olarak anlamlı şekilde artırmaktadır.")
else:
    print(" Sonuç: H0 reddedilemez. İlacın etkisi henüz anlamlı düzeyde kanıtlanamamıştır.")
```

---
## 📝 Özet ve Çıkarımlar
1. **Z-Testi:** Popülasyon varyansı $\sigma$ kesin bilindiğinde ve örneklem yeterince büyük ($n \ge 30$) olduğunda tercih edilir.
2. **T-Testi:** Popülasyon standart sapması bilinmediğinde doğrudan örneklemin standart sapması $s$ alınarak serbestlik derecesi ($df = n - 1$) üzerinden çalışır.
3. **Karar Kuralı:** $p < 0.05$ olduğunda sıfır hipotezi reddedilir, değişim anlamlıdır.
