# 🎲 Ki-Kare Bağımsızlık Testi (Chi-Square Test of Independence)

Ki-Kare ($\chi^2$) Bağımsızlık Testi, **iki kategorik değişken** arasında istatistiksel olarak anlamlı bir ilişki (bağımlılık) olup olmadığını tespit etmek için kullanılan parametrik olmayan (non-parametric) bir hipotez testidir.

---

## 🎯 1. Teorik Altyapı ve Formüller

### 📐 Formül:
$$\chi^2 = \sum_{i=1}^{k} \frac{(O_i - E_i)^2}{E_i}$$

- **$O_i$ (Observed):** Gözlemlenen gerçek frekans sayısı
- **$E_i$ (Expected):** İki değişken bağımsız olsaydı beklenen teorik frekans sayısı
- **Serbestlik Derecesi ($df$):** $df = (\text{Satır Sayısı} - 1) \times (\text{Sütun Sayısı} - 1)$

### ⚖️ Hipotezler:
- **$H_0$ (Sıfır Hipotezi):** İki kategorik değişken birbirinden **bağımsızdır** (Aralarında ilişki yoktur).
- **$H_1$ (Alternatif Hipotez):** İki kategorik değişken birbirinden **bağımsız değildir** (Aralarında anlamlı bir ilişki/bağlantı vardır).

### ⚠️ Geçerlilik ve Varsayımlar (Cochran Kuralları):
1. Veriler nitel / kategorik olmalıdır.
2. Gözlemler birbirinden bağımsız olmalıdır.
3. Kontenjans tablosundaki beklenen frekansların ($E_i$) en az %80’i 5 veya daha büyük olmalıdır. Hiçbir hücre 1’den küçük olamaz.


---
## 🏬 2. Senaryo 1: Müşteri Memnuniyeti ve Ziyaret Zamanı (Hafta İçi / Hafta Sonu)

Müşteri memnuniyet seviyeleri (Memnun / Memnun Değil) ile mağazayı ziyaret ettikleri gün türü (Hafta İçi / Hafta Sonu) arasında ilişki var mıdır?


```python
import pandas as pd
from scipy.stats import chi2_contingency

# Kontenjans Tablosu (Satırlar: [Memnun, Memnun Değil], Sütunlar: Hafta İçi, Hafta Sonu)
data = {
    "Hafta İçi": [30, 10],
    "Hafta Sonu": [20, 40]
}

df_satisfaction = pd.DataFrame(data, index=["Memnun", "Memnun Değil"])
print("--- Gözlenen Frekanslar (Observed) ---")
print(df_satisfaction)

chi2, p_value, dof, expected = chi2_contingency(df_satisfaction)

print(f"
Ki-Kare Değeri       : {chi2:.4f}")
print(f"p-değeri             : {p_value:.6f}")
print(f"Serbestlik Derecesi  : {dof}")
print("
--- Beklenen Frekanslar (Expected) ---")
print(pd.DataFrame(expected, index=df_satisfaction.index, columns=df_satisfaction.columns))

alpha = 0.05
if p_value < alpha:
    print("
 Sonuç: H0 reddedilir! Hafta içi / hafta sonu gelme durumu ile müşteri memnuniyeti arasında anlamlı bir ilişki vardır.")
else:
    print("
 Sonuç: H0 reddedilemez. Değişkenler bağımsızdır.")
```

## 🍎 3. Senaryo 2: Cinsiyet ile Ürün Tercihi Arasındaki İlişki (`pd.crosstab`)


```python
import pandas as pd
from scipy.stats import chi2_contingency

df = pd.DataFrame({
    "Cinsiyet": ["Kadın", "Erkek", "Kadın", "Kadın", "Erkek", "Kadın", "Erkek", "Erkek", "Erkek", "Kadın"],
    "Ürün": ["Elma", "Elma", "Elma", "Elma", "Armut", "Armut", "Armut", "Elma", "Armut", "Elma"]
})

# Çapraz Tablo (Contingency Table)
table = pd.crosstab(df["Cinsiyet"], df["Ürün"])
print("--- Çapraz Tablo ---")
print(table)

stat, p_value, dof, expected = chi2_contingency(table)

print(f"
Ki-Kare İstatistiği: {stat:.4f}")
print(f"p-değeri           : {p_value:.4f}")

if p_value < 0.05:
    print(" Sonuç: Cinsiyet ile meyve tercihi arasında istatistiksel olarak anlamlı bir ilişki vardır.")
else:
    print(" Sonuç: Cinsiyet ile meyve tercihi arasında anlamlı bir ilişki bulunamamıştır (Bağımsızdırlar).")
```

## 🚬 4. Senaryo 3: Gruplar ve Sigara Kullanım Alışkanlığı


```python
import pandas as pd
from scipy.stats import chi2_contingency

df_smoke = pd.DataFrame({
    "Sigara İçiyor": [90, 30],
    "Sigara İçmiyor": [60, 120]
}, index=["Grup 1", "Grup 2"])

print("--- Gözlenen Veriler ---")
print(df_smoke)

chi2, p_value, dof, expected = chi2_contingency(df_smoke)

print(f"
Ki-Kare İstatistiği: {chi2:.4f}")
print(f"p-değeri           : {p_value:.8f}")

if p_value < 0.05:
    print(" Sonuç: H0 reddedilir. Grupların sigara tüketim oranları arasında istatistiksel olarak çok güçlü bir ilişki vardır.")
else:
    print(" Sonuç: Gruplar arasında sigara kullanımı açısından fark yoktur.")
```
