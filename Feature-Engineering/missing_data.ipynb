# 🩹 Eksik Veri Analizi ve Doldurma Yöntemleri (Handling Missing Data)

Gerçek dünya veri setlerinde eksik veri (missing data / NaN) kaçınılmaz bir gerçektir. 
Eksik değerleri doğru analiz etmeden model eğitmek, yanıltıcı sonuçlara yol açar.

---

## 📌 Eksik Veri Türleri:
1. **MCAR (Missing Completely at Random):** Eksiklik tamamen rastgeledir; diğer hiçbir sütunla ilgisi yoktur.
2. **MAR (Missing at Random):** Eksiklik başka bir gözlemlenebilir değişkenle ilişkilidir (örn: kadınların gelir alanını boş bırakma eğilimi).
3. **MNAR (Missing Not at Random):** Eksiklik doğrudan değişkenin kendi değeriyle ilgilidir (örn: çok yüksek gelirli kişilerin gelirlerini belirtmemesi).

### 🛠️ Başlıca Çözüm Stratejileri:
- **Satır / Sütun Silme (Deletion):** Sütunda %70'ten fazla eksiklik varsa sütun silinebilir; az sayıda eksik varsa satırlar düşürülebilir (`dropna`).
- **Ortalama (Mean) / Medyan (Median) ile Doldurma:** Simetrik dağılımlarda ortalama, çarpık (skewed) dağılımlarda medyan tercih edilir.
- **Mod (Mode) ile Doldurma:** Kategorik verilerde en sık tekrar eden değer kullanılır.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Klasik Titanic Veri Seti
df_titanic = sns.load_dataset("titanic")
df_titanic.head()
```

### 🔍 1. Eksik Değerlerin Tespiti ve Dağılımı


```python
# Her sütundaki eksik veri sayısı ve yüzdesi
null_counts = df_titanic.isnull().sum()
null_percent = (null_counts / len(df_titanic)) * 100

missing_df = pd.DataFrame({"Eksik Adedi": null_counts, "Yüzde (%)": null_percent.round(2)})
missing_df[missing_df["Eksik Adedi"] > 0].sort_values(by="Yüzde (%)", ascending=False)
```

### 📈 2. Sürekli Değişken Analizi: `age` Sütunu
`age` sütununun dağılımına bakarak Ortalama mı Medyan mı kullanacağımıza karar veririz.


```python
plt.figure(figsize=(9, 4))
sns.histplot(df_titanic["age"], kde=True, color="teal")
plt.title("Yaş (Age) Değişkeni Dağılımı")
plt.xlabel("Yaş")
plt.ylabel("Frekans")
plt.show()

# Medyan ve Ortalama Değerler
age_mean = df_titanic["age"].mean()
age_median = df_titanic["age"].median()
print(f"Yaş Ortalaması: {age_mean:.2f}, Yaş Medyanı: {age_median:.2f}")

# Medyan ile Doldurma (Imputation)
df_titanic["age_imputed_median"] = df_titanic["age"].fillna(age_median)
print(f"Eksik yaş sayısı (önce): {df_titanic['age'].isnull().sum()} -> (sonra): {df_titanic['age_imputed_median'].isnull().sum()}")
```

### 🚢 3. Yüksek Eksiklik İçeren Kategorik Değişken: `deck` Sütunu
`deck` sütununun %77'si eksiktir. Bu gibi durumlarda ya 'Missing' adında yeni bir kategori oluşturulur ya da sütun düşürülür.


```python
print("Deck Sütunu Değer Dağılımı:")
print(df_titanic["deck"].value_counts(dropna=False))

plt.figure(figsize=(8, 4))
sns.countplot(x=df_titanic["deck"].astype(str), palette="viridis")
plt.title("Deck Sütunu Frekans Grafiği")
plt.show()
```
