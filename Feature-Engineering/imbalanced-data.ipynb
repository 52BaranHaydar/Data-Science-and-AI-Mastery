# ⚖️ Dengesiz Veri Kümeleri ile Mücadele (Imbalanced Datasets)

Dolandırıcılık tespiti (Fraud Detection - %99 normal, %1 sahtekarlık), kanser teşhisi veya müşteri terk (Churn) problemlerinde hedef sınıf oranları aşırı dengesizdir.
Böyle bir veride model herkese "Normal" derse %99 doğruluk (Accuracy) elde eder ama hiçbir sahtekarlığı yakalayamaz (Accuracy Paradox).

---

## 🛠️ Çözüm Yolları:
1. **Downsampling (Azaltarak Dengeleme):** Çoğunluk sınıfından rastgele veri silerek azınlık sınıfı sayısına indirir. *(Veri kaybı riski taşır).*
2. **Upsampling (Çoğaltarak Dengeleme):** Azınlık sınıfındaki mevcut verileri kopyalar. *(Aşırı öğrenme - Overfitting riski taşır).*
3. **Sentetik Örnekleme (SMOTE):** Azınlık sınıfı verileri arasında interpolasyon yaparak yepyeni sentetik örnekler üretir.


```python
import pandas as pd
import numpy as np

np.random.seed(42)

# %90 Sınıf 0, %10 Sınıf 1 olan sentetik veri kümesi
df_class_0 = pd.DataFrame({
    "feature_1": np.random.normal(loc=0, scale=1, size=900),
    "feature_2": np.random.normal(loc=0, scale=1, size=900),
    "target": [0] * 900
})

df_class_1 = pd.DataFrame({
    "feature_1": np.random.normal(loc=2, scale=1, size=100),
    "feature_2": np.random.normal(loc=2, scale=1, size=100),
    "target": [1] * 100
})

df_total = pd.concat([df_class_0, df_class_1], ignore_index=True)
print("Orijinal Sınıf Dağılımı:")
print(df_total["target"].value_counts())
```

### 📉 1. Downsampling (Çoğunluk Sınıfını Azaltma)


```python
from sklearn.utils import resample

df_minority = df_total[df_total["target"] == 1]
df_majority = df_total[df_total["target"] == 0]

# Çoğunluk sınıfından azınlık sayısı kadar örnek seçme
df_majority_downsampled = resample(
    df_majority,
    replace=False,
    n_samples=len(df_minority),
    random_state=42
)

df_downsampled = pd.concat([df_minority, df_majority_downsampled])
print("Downsampling Sonrası Sınıf Dağılımı:")
print(df_downsampled["target"].value_counts())
```
