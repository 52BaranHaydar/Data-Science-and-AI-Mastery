# 🏷️ Kategorik Veri Dönüşümü: Label & Ordinal Encoding

Makine öğrenmesi algoritmaları yalnızca matematiksel matrisler ve nümerik değerlerle çalışır. 
Kategorik metin verilerini sayısal temsillere dönüştürmek gerekir.

---

## 📌 Label Encoding vs Ordinal Encoding

| Yöntem | Ne Zaman Kullanılır? | Örnek |
| :--- | :--- | :--- |
| **Ordinal Encoding** | Kategoriler arasında **doğal bir hiyerarşi/üstünlük sıralaması** olduğunda | Beden (`small < medium < large < xlarge`), Eğitim (`İlkokul < Lise < Lisans < Yüksek Lisans`) |
| **Label Encoding** | Genellikle **hedef değişken (Target $y$)** üzerinde ikili veya çok sınıflı etiketleme için | `[Erkek, Kadın] -> [0, 1]`, `[Kedi, Köpek, Kuş] -> [0, 1, 2]` |

⚠️ **Dikkat:** Bağımsız değişkenlerde (X) sıralama içermeyen kategorilere Label Encoding uygulanırsa model yapay bir büyüklük-küçüklük algılayabilir (örn: Yeşil=2, Mavi=0 ise model yeşili maviden büyük sanabilir).


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

df_color = pd.DataFrame({
    "color": ["red", "green", "red", "blue", "green", "red"]
})

label_encoder = LabelEncoder()
df_color["encoded_color"] = label_encoder.fit_transform(df_color["color"])

print("--- Renkler Label Encoding ---")
print(df_color)
print("Sınıf Sıralaması:", label_encoder.classes_)
```

### 📏 Ordinal Encoding: Sıralı Hiyerarşik Veriler


```python
from sklearn.preprocessing import OrdinalEncoder

df_size = pd.DataFrame({
    "size": ["small", "medium", "large", "xlarge", "medium", "small", "large"]
})

# Sıralamayı küçükten büyüğe elle tanımlıyoruz
ordered_categories = [["small", "medium", "large", "xlarge"]]
ordinal_encoder = OrdinalEncoder(categories=ordered_categories)

df_size["size_num"] = ordinal_encoder.fit_transform(df_size[["size"]]).astype(int)

print("--- Bedenler Ordinal Encoding ---")
print(df_size)
```

### 🏙️ Çoklu Sütunlarda Bağımsız Encoder Kullanımı


```python
df_person = pd.DataFrame({
    "sehir": ["İstanbul", "Ankara", "İzmir", "İstanbul"],
    "cinsiyet": ["Erkek", "Kadın", "Kadın", "Erkek"]
})

# Her sütun için ayrı bir encoder tanımlanmalıdır
sehir_encoder = LabelEncoder()
df_person["encoded_sehir"] = sehir_encoder.fit_transform(df_person["sehir"])

cinsiyet_encoder = LabelEncoder()
df_person["encoded_cinsiyet"] = cinsiyet_encoder.fit_transform(df_person["cinsiyet"])

print("--- Bağımsız Sütun Dönüşümleri ---")
print(df_person)
```
