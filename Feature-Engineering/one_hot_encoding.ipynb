# 🔘 One-Hot Encoding (Nominal Kategorik Veriler)

Nominal verilerde (aralarında hiyerarşi/büyüklük ilişkisi olmayan kategoriler; örn: Kırmızı, Yeşil, Mavi veya Şehirler), Label Encoding modelde yapay ağırlık hatalarına yol açar.
**One-Hot Encoding**, her kategori için $0$ ve $1$'lerden oluşan ayrı bir sütun oluşturur.

---

## ⚠️ Dummy Variable Trap (Kukla Değişken Tuzağı)
Bir sütundaki değerler diğer sütunların toplamından tam olarak tahmin edilebiliyorsa (tam çoklu doğrusal bağlantı - multicollinearity), model matris tersi alırken tekil matris (singular matrix) hatası oluşur.
- Çözüm: $k$ kategori varsa $k - 1$ sütun tutulur (`drop='first'`).


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df_color = pd.DataFrame({
    "color": ["red", "green", "red", "blue", "green", "red"]
})

encoder = OneHotEncoder(sparse_output=False)
encoded_array = encoder.fit_transform(df_color[["color"]])
encoded_df = pd.DataFrame(encoded_array, columns=encoder.get_feature_names_out(["color"]))

final_df = pd.concat([df_color, encoded_df], axis=1)
print("--- Scikit-Learn OneHotEncoder Sonucu ---")
print(final_df)
```

### 🍽️ Gerçek Veri Örneği: Seaborn `tips` Veri Seti


```python
import seaborn as sns

df_tips = sns.load_dataset("tips")

encoder_tips = OneHotEncoder(drop="first", sparse_output=False)
sex_encoded = encoder_tips.fit_transform(df_tips[["sex"]])
sex_encoded_df = pd.DataFrame(sex_encoded, columns=encoder_tips.get_feature_names_out(["sex"]))

tips_transformed = pd.concat([df_tips[["total_bill", "tip", "sex"]], sex_encoded_df], axis=1)
tips_transformed.head()
```
