# 🎯 Hedefe Yönelik Kodlama (Target Guided Mean Encoding)

Bir kategorik sütunda **onlarca veya yüzlerce farklı kategori (yüksek kardinalite - high cardinality)** varsa, One-Hot Encoding sütun patlamasına (Curse of Dimensionality) yol açar.

**Target Guided Encoding**, her kategoriyi hedef değişkenin ($y$) o kategorideki ortalama değeriyle eşleştirir.

---

### 💡 Avantajları:
1. Sütun sayısını artırmaz (1 sütun kalır).
2. Model ile hedef değişken arasında monotonik (doğrusal) bir ilişki kurar.


```python
import pandas as pd

# Şehirler ve Ortalama Ev / Oda Fiyatları
df_realestate = pd.DataFrame({
    "city": ["New York", "London", "Paris", "Tokyo", "New York", "Paris"],
    "price": [200, 150, 300, 250, 180, 320]
})

print("Orijinal Veri:")
print(df_realestate)

# Her şehrin ortalama fiyatını hesaplayıp haritalama sözlüğü oluşturuyoruz
city_mean_price = df_realestate.groupby("city")["price"].mean().round(2).to_dict()
print("
Şehir Başına Ortalama Hedef (Price):", city_mean_price)

# Sütunu ortalama değer ile dönüştürme
df_realestate["city_encoded"] = df_realestate["city"].map(city_mean_price)
print("
Target Guided Encoded Veri:")
print(df_realestate)
```

### 🍽️ Restoran Verisinde Öğün Zamanı (`time`) ve Ortalama Hesap (`total_bill`)


```python
import seaborn as sns

df_tips = sns.load_dataset("tips")

# Öğüne göre ortalama hesap tutarı
time_mean_bill = df_tips.groupby("time", observed=True)["total_bill"].mean().round(2).to_dict()
print("Zaman Başına Ortalama Hesap:", time_mean_bill)

df_tips["time_encoded"] = df_tips["time"].map(time_mean_bill)
df_tips[["total_bill", "time", "time_encoded"]].head()
```
