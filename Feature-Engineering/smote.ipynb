# 🧬 SMOTE: Synthetic Minority Over-sampling Technique

SMOTE, basitçe mevcut verileri kopyalamak yerine, **k-En Yakın Komşu (k-NN)** algoritmasını kullanarak azınlık sınıfı noktaları arasındaki doğru boyunca **yeni sentetik örnekler** üretir.

---

### 📐 SMOTE Çalışma Prensibi:
1. Azınlık sınıfından bir $x_i$ noktası seçilir.
2. $x_i$'nin azınlık sınıfı içindeki $k$ en yakın komşusu bulunur.
3. Bu komşulardan rastgele biri ($x_{zi}$) seçilir.
4. İki nokta arasındaki doğru parçası üzerinde rastgele bir noktada yeni sentetik veri üretilir:
$$x_{\text{yeni}} = x_i + \lambda \times (x_{zi} - x_i), \quad \lambda \in [0, 1]$$


```python
from sklearn.datasets import make_classification
import pandas as pd
import matplotlib.pyplot as plt

# %90'a %10 Dengesiz Veri Üretimi
X, y = make_classification(
    n_samples=1000,
    n_features=2,
    n_redundant=0,
    n_clusters_per_class=1,
    weights=[0.90],
    random_state=12
)

final_df = pd.DataFrame(X, columns=["f1", "f2"])
final_df["target"] = y

print("Orijinal Sınıf Frekansları:")
print(final_df["target"].value_counts())

plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="bwr", alpha=0.7, edgecolors="k")
plt.title("Orijinal Dengesiz Dağılım (90% Mavi vs 10% Kırmızı)")
plt.xlabel("Özellik 1 (f1)")
plt.ylabel("Özellik 2 (f2)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

### 🧪 SMOTE ile Sentetik Örnekleme ve Sonuç


```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(final_df[["f1", "f2"]], final_df["target"])

balanced_df = pd.DataFrame(X_resampled, columns=["f1", "f2"])
balanced_df["target"] = y_resampled

print("SMOTE Sonrası Sınıf Dağılımı:")
print(balanced_df["target"].value_counts())

plt.figure(figsize=(7, 5))
plt.scatter(balanced_df["f1"], balanced_df["f2"], c=balanced_df["target"], cmap="bwr", alpha=0.6, edgecolors="k")
plt.title("SMOTE Sonrası Mükemmel Dengeli Dağılım (50% vs 50%)")
plt.xlabel("Özellik 1 (f1)")
plt.ylabel("Özellik 2 (f2)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
