# 🎯 Aykırı Değer Analizi ve IQR Yöntemi (Handling Outliers)

Veri setlerinde genel dağılımın çok dışında kalan gözlemlere **aykırı değer (outlier)** denir. 
Makine öğrenmesi modellerinin (özellikle Doğrusal Regresyon, Lojistik Regresyon, k-NN, K-Means gibi mesafe ve varyansa dayalı modellerin) performansını ciddi şekilde bozar.

---

## 📐 Çeyrekler Açıklığı (IQR - Interquartile Range) Yöntemi
Veri sıralandıktan sonra:
- $Q_1$ (%25'lik Dilim - 1. Çeyrek)
- $Q_2$ (Medyan - %50'lik Dilim)
- $Q_3$ (%75'lik Dilim - 3. Çeyrek)
- $\text{IQR} = Q_3 - Q_1$

### 🚪 Çit (Fence) Sınırları:
$$\text{Alt Sınır (Lower Fence)} = Q_1 - 1.5 \times \text{IQR}$$
$$\text{Üst Sınır (Upper Fence)} = Q_3 + 1.5 \times \text{IQR}$$

Bu sınırların dışında kalan her nokta **aykırı değer (outlier)** kabul edilir.


```python
import numpy as np

# Öğrenci Sınav Notları Örnek Veri Seti
list_marks = [45, 32, 45, 75, 89, 54, 32, 89, 90, 87, 67, 55, 48, 74, 72, 77, 99]

# 5 Sayı Özeti (Minimum, Q1, Medyan, Q3, Maksimum)
minimum, q1, median, q3, maximum = np.quantile(list_marks, [0, 0.25, 0.5, 0.75, 1.0])

print(f"Minimum: {minimum}")
print(f"Q1 (%25) : {q1}")
print(f"Medyan (%50): {median}")
print(f"Q3 (%75) : {q3}")
print(f"Maksimum: {maximum}")
```

### 🔍 Çit Hesaplama ve Aykırı Değer Filtreleme


```python
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
higher_fence = q3 + 1.5 * iqr

print(f"IQR (Q3 - Q1) : {iqr}")
print(f"Alt Çit (Lower Fence) : {lower_fence:.2f}")
print(f"Üst Çit (Upper Fence) : {higher_fence:.2f}")

# Aykırı değerleri tespit etme
outliers = [x for x in list_marks if x < lower_fence or x > higher_fence]
cleaned_data = [x for x in list_marks if lower_fence <= x <= higher_fence]

print(f"
Aykırı Değerler : {outliers}")
print(f"Temizlenmiş Veri: {cleaned_data}")
```

### 📊 Kutu Grafiği (Boxplot) ile Görselleştirme


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 4))
sns.boxplot(x=list_marks, color="skyblue")
plt.title("Sınav Notları Dağılımı ve Aykırı Değerler (Boxplot)")
plt.xlabel("Notlar")
plt.axvline(lower_fence, color="red", linestyle="--", label=f"Alt Çit ({lower_fence:.1f})")
plt.axvline(higher_fence, color="red", linestyle="--", label=f"Üst Çit ({higher_fence:.1f})")
plt.legend()
plt.show()
```
