```python
# Bir metin sütunu olan yorumlar sütunundaki her sütunundaki her bir metnin uzunkluğunu (karakter cinsinden)
# yorum_uzunluğu adlı yeni bir sutun ekleyen python kodunu yazınız

import pandas as pd

df = pd.DataFrame({""
        "yorumlar" : ["Harika ürün!","Beğenmedim","Fena Değil"]
})

df["yorum_uzunluğu"] = df["yorumlar"].apply(len)
df
```

```python
# Bir veri kümesinde sehir adında kategorik bir sutun var
# Bu sütunu one-hot encoding yöntemi ile sayısal değerlere dönüştüren kodu yazınız.
import pandas as pd

df = pd.DataFrame({
    "sehir": ["Ankara","İstanbul","İzmir","Ankara"]
})

df_encoded = pd.get_dummies(df, columns = ["sehir"])
df_encoded
```
